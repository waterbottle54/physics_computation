```python
import numpy as np
import random as rand
import pandas as pd
import math
```

```python
def calc_tag_position(dist0, distx, disty, x_anchorx, y_anchory):
    d0, dx, dy = dist0, distx, disty
    xa, ya = x_anchorx, y_anchory
    d02, dx2, dy2 = d0**2, dx**2, dy**2
    x = (d0**2 - dx**2)/(2*xa) + (xa/2)
    y = (d0**2 - dy**2)/(2*ya) + (ya/2)
    z = np.sqrt(d02 - x**2 - y**2)
    return x, y, z

def calc_error(actual, calculated):
    return np.where(actual != 0, np.abs((actual - calculated)/actual) * 100, 0)

# 앵커 2개 위치 (나머지 1개는 원점)
x_anchorx, y_anchory = 1.0, 1.0

# 태그 10개 위치 생성
n_tags = 10
x_actual = np.random.rand(n_tags) * 10 - 5
y_actual = np.random.rand(n_tags) * 10 - 5
z_actual = np.random.rand(n_tags) * 3

# 태그와 앵커 3개간 거리 계산 (UWB 대신)
# 딜레이 0
dist0 = np.sqrt(x_actual**2 + y_actual**2 + z_actual**2)
distx = np.sqrt((x_actual - x_anchorx)**2 + y_actual**2 + z_actual**2)
disty = np.sqrt(x_actual**2 + (y_actual - y_anchory)**2 + z_actual**2)

# 태그 위치 계산 - 거리 3개 기반
x_calc, y_calc, z_calc = calc_tag_position(dist0, distx, disty, x_anchorx, y_anchory)

# 오차 계산
error_x = np.where(x_actual != 0, calc_error(x_actual, x_calc), 0)
error_y = np.where(y_actual != 0, calc_error(y_actual, y_calc), 0)
error_z = np.where(z_actual != 0, calc_error(z_actual, z_calc), 0)

# df로 출력
df = pd.DataFrame({
    'x_actual': x_actual,
    'x_calcul': x_calc,
    'y_actual': y_actual,
    'y_calcul': y_calc,
    'z_actual': z_actual,
    'z_calcul': z_calc,
    'err_x(%)': error_x,
    'err_y(%)': error_y,
    'err_z(%)': error_z
})

pd.set_option('display.float_format', '{:.15f}'.format)
display(df.T)
```
